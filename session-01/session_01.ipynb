```python
import os
from google.colab import userdata
```

```python
os.environ["ANTHROPIC_API_KEY"] = userdata.get("ANTHROPIC_API_KEY")
```

```python
from anthropic import Anthropic
```

```python
client = Anthropic()
```

```python
response = client.messages.create(
    model="claude-opus-5-5",
    max_tokens=16000,
    messages=[{"role": "user", "content": "¿Cuánto se tarda en desarrollar una app de reservas para un restaurante?"}]
)
```

```python
response
```

```python
print(response.content[1].text)
```

```python
response_2 = client.messages.create(
    model="claude-opus-5-5",
    max_tokens=16000,
    messages=[{"role": "user", "content": "¿Cuánto se tarda en desarrollar una app de reservas para un restaurante?"}],
    system="Eres un experto en desarrollo de software, respondes con argumentos técnicos y detallados, para dar seguridad y generar confianza"
)
```

```python
print(response_2.content[1].text)
```

```python
model = response_2.model
```

```python
input_tokens = response_2.usage.input_tokens
```

```python
output_tokens = response_2.usage.output_tokens
```

```python
print(f"- Model:{model}\n- Input Tokens:{input_tokens}\n- Output Tokens:{output_tokens}")
```
